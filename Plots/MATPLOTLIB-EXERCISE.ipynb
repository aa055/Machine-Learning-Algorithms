```python
fig = plt.figure()

# Add set of axes to figure
axes = fig.add_axes([0, 0, 1, 1]) # left, bottom, width, height (range 0 to 1)
# Plot on that set of axes
axes.plot(x, y, 'b')
axes.set_xlabel('X') 
axes.set_ylabel('Y')
axes.set_title('Title')

yticks = [0,50,100,150,200]
axes.set_yticks(yticks)

axes.set_ylim([0, 200])
axes.set_xlim([0, 100])
```

## Exercise 2
** Create a figure object and put two axes on it, ax1 and ax2. Located at [0,0,1,1] and [0.2,0.5,.2,.2] respectively.**

```python
# Creates blank canvas
fig = plt.figure()

axes1 = fig.add_axes([0, 0, 1, 1])
axes2 = fig.add_axes([0.2, 0.5, 0.2, 0.2]) 
```

** Now plot (x,y) on both axes with labels. And call your figure object to show it.**

```python
# Larger Figure Axes 1
axes1.plot(x, y, 'r')
axes1.set_xlabel('X1')
axes1.set_ylabel('Y1')

# Insert Figure Axes 2
axes2.plot(x, y, 'r')
axes2.set_xlabel('X2')
axes2.set_ylabel('Y2')

fig
```

## Exercise 3

** Create the plot below by adding two axes to a figure object at [0,0,1,1] and [0.2,0.5,.4,.4]**

```python
# Creates blank canvas
fig2 = plt.figure()

axes1 = fig2.add_axes([0, 0, 1, 1])
axes2 = fig2.add_axes([0.2, 0.5, 0.4, 0.4]) 
```

** Now use x,y, and z arrays to recreate the plot below. Notice the xlimits and y limits on the inserted plot:**

```python
# Larger Figure Axes 1
axes1.plot(x, z, 'r')
axes1.set_xlabel('X')
axes1.set_ylabel('Z')

# Insert Figure Axes 2
axes2.plot(x, y, 'r')
axes2.set_xlabel('X')
axes2.set_ylabel('Y')
axes2.set_title('zoom')

axes2.set_ylim([30, 50])
axes2.set_xlim([20, 22])

fig2
```

## Exercise 4

** Use plt.subplots to create the plot below.**

```python
fig4,axes4 = plt.subplots(nrows = 1,ncols=2)
```

** Now plot (x,y) and (x,z) on the axes. Play around with the linewidth and style**

```python
axes4[0].plot(x, y, 'b--', lw=4)
axes4[1].plot(x, z, 'r', lw=4)
fig4.tight_layout()
fig4
```

** See if you can resize the plot by adding the figsize() argument in plt.subplots() by copying and pasting your previous code.**

```python
fig5,axes5 = plt.subplots(nrows = 1,ncols=2, figsize=(10,2))

axes5[0].plot(x, y, 'b', lw=4)
axes5[1].plot(x, z, 'r--', lw=4)
fig5.tight_layout()
```
